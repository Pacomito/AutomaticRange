```python
from tifffile import imread, imwrite
import torch
import os
import numpy as np
from AutomaticRange.models import AutomaticRangeNet
from AutomaticRange.smooth import predict_range
```

```python
checkpoint_path = os.path.join(
    os.path.dirname(os.path.dirname("../../")),
    "checkpoints",
    "training_set_processed_CD4_nsamp_20_ntile_5_07102025_automatic_range.pt"
)

# Load trained model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = AutomaticRangeNet().to(device)
model.load_state_dict(torch.load(checkpoint_path, map_location=device))
model.eval()
```

```python
dapi = imread('/media/localadmin/EGlab20TB-5/TMA_final_run_01/input_processed/AD145_TMA1/DAPI.tiff')
cd4 = imread('/media/localadmin/EGlab20TB-5/TMA_final_run_01/input_processed/AD145_TMA1/CD4.tiff')

normalized_image, min_interp, max_interp  = predict_range(dapi=dapi, marker=cd4, model=model, device = device) 
imwrite('../data/AD145_TMA1_CD4_norm.tiff', normalized_image.astype(np.float16))
```
